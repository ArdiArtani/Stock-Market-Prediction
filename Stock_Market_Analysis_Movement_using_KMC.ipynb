```python
# import libraries for Machine Learning
from sklearn.pipeline import make_pipeline
from sklearn.cluster import KMeans
from sklearn.preprocessing import Normalizer
from sklearn.decomposition import PCA

# initalize Normalizer
normalizer = Normalizer()
fit = normalizer.fit_transform(volumes)
normalizer = Normalizer()
KMC = KMeans(n_clusters=3, max_iter=1000)
pipeline = make_pipeline(normalizer,KMC)
pipeline.fit(volumes)

# making cluster label for the df
labels = pipeline.predict(volumes)
df = pd.DataFrame({'labels': labels, 'tickers': tickers})
PCA_data = PCA(n_components = 2).fit_transform(fit)

# run Kmeans on PCA_data
KMC = KMeans(n_clusters=3)
KMC.fit(PCA_data)
labels = KMC.predict(PCA_data)

# Make df for labels and tickers
df = pd.DataFrame({'labels': labels, 'tickers': tickers})

# ploting the decision boundary
X_min, X_max = PCA_data[:, 0].min() - 1, PCA_data[:,0].max() + 1
Y_min, Y_max = PCA_data[:, 1].min() - 1, PCA_data[:,1].max() + 1
X, Y = np.meshgrid(np.arange(X_min, X_max, 0.01), np.arange(Y_min, Y_max, 0.01))

# Calculate Z-score index
Z = KMC.predict(np.c_[X.ravel(), Y.ravel()]).reshape(X.shape)

# plot figure
plt.clf()
plt.figure(figsize=(10,10))
plt.imshow(Z, interpolation='nearest', extent = (X.min(), X.max(), Y.min(), Y.max()), cmap = plt.cm.Pastel2, aspect = 'auto', origin='lower')
plt.plot(PCA_data[:, 0], PCA_data[:, 1], 'k.', markersize=5)
plt.title('KMC on Stock Market Movement')
plt.xlim(X_min, X_max)
plt.ylim(Y_min, Y_max)
plt.show()
plt.savefig('KMC.png')
```

```python
# Helper Functions
def getPrice(ticker, time):
    df = stocks[ticker]
    q = f'select * from df WHERE Date = "{time}" '
    df = psql.sqldf(q)
    return df["Close"][0]
```

```python
# Import Libraries
import numpy as np
import pandas as pd
import copy
import statistics
import pandasql as psql
from pandas_datareader import data as pdr
import yfinance as yf
from datetime import date
from datetime import datetime
from datetime import timedelta
yf.pdr_override() # <== that's all it takes :-)

# Principal amount in USD for investing
principal = 10000

# Initailizing portfolio
portfolio = {
    "AAPL": 0,
    "MSFT": 0,
}

# Timeframe
start_date = '2019-01-01'
end_date = '2019-12-31'
dt_start = datetime.strptime(start_date, '%Y-%m-%d')

# Get stocks DF for the timeframe
stocks = copy.copy(portfolio)
for i in portfolio:
    data = pdr.get_data_yahoo(i, start=start_date, end=end_date)
    q = f"SELECT Date, Close FROM data;"
    df = psql.sqldf(q)
    stocks[i] = df
# print(stocks)

# Calculate a 50/50 auto-balanced portfolio
for i in portfolio:
    shares = principal/ len(portfolio)/ getPrice(i, (str(dt_start + timedelta(days=1))[:10] + " 00:00:00.000000"))
    portfolio[i] = shares
# print(portfolio)
```

```python
# Make a Simulation for the portfolio with the given time frame
trading_days = stocks[list(stocks.keys())[0]]['Date']
for i in trading_days:
    df = pd.DataFrame()
    Ticker = list(stocks.keys())
    df['Ticker'] = list(stocks.keys())

    Quantity = []
    for j in portfolio:
        Quantity.append(portfolio[j])
    df['Quantity'] = Quantity

    Value = []
    for j in range(len(Quantity)):
        val = Quantity[j] * getPrice(Ticker[j], i)
        Value.append(val)
    df["Value"] = Value

    Percentage = []
    for j in range(len(Value)):
        percentage = Value[j]/sum(Value)
        Percentage.append(percentage)
    df["Percentage"] = Percentage

    # print statements
#     print(i)
#     print(df)
    total_value = round(sum(Value), 2)
before = total_value
#     print(f"\nTotal: {total_value}")
#     print('__________________________________')
print(f"Portfolio Value: {total_value}")
```

```python
# Make a Simulation for the portfolio with the given time frame with Auto-Rebalance enabled
trading_days = stocks[list(stocks.keys())[0]]['Date']
for i in trading_days:
    df = pd.DataFrame()
    Ticker = list(stocks.keys())
    df['Ticker'] = list(stocks.keys())

    Quantity = []
    for j in portfolio:
        Quantity.append(portfolio[j])
    df['Quantity'] = Quantity

    Value = []
    for j in range(len(Quantity)):
        val = Quantity[j] * getPrice(Ticker[j], i)
        Value.append(val)
    df["Value"] = Value

    Percentage = []
    for j in range(len(Value)):
        percentage = Value[j]/sum(Value)
        Percentage.append(percentage)
    df["Percentage"] = Percentage

    # Rebalance (sell everything and buy back in with the original ratio)
    if df["Percentage"][0] != statistics.mean(df["Percentage"]):
        for j in portfolio:
            shares = sum(Value)/ len(portfolio)/ getPrice(j, i)
            portfolio[j] = shares

        Quantity = []
        for j in portfolio:
            Quantity.append(portfolio[j])
        df['Quantity'] = Quantity

        Value = []
        for j in range(len(Quantity)):
            val = Quantity[j] * getPrice(Ticker[j], i)
            Value.append(val)
        df["Value"] = Value

        Percentage = []
        for j in range(len(Value)):
            percentage = Value[j]/sum(Value)
            Percentage.append(percentage)
        df["Percentage"] = Percentage

    # print statements
#     print(i)
#     print(df)
    total_value = round(sum(Value), 2)
    after = total_value
#     print(f"\nTotal: {total_value}")
#     print('__________________________________')
print(f"Portfolio Value Auto-Rebalance: {total_value}")
print(f"Additional Gains: {after - before}")
print(f"Additional Gains Percentage: {(after/before * 100) -100  }")
```
